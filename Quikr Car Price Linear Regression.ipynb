## Quality + Problem to be Resolved
- year has many non-year values
- year dtype from object to integer
- price has ask for price
- price dtype from object to integer
- kms_driven object to integer
- kms_driven has nan values
- fuel_type has nan values
- keep first three words of name

# Cleaning

```python
backup = car.copy()
```

```python
#Right Now, year dtype is object
car.info()
```

```python
car['year'] = car['year'].astype(str) #Ensures string dtype first
car = car[car['year'].str.isnumeric()] #Filter numeric values and keep only true values
car['year'] = pd.to_numeric(car['year']) # Convert into integer
```

```python
car.info()
```

```python
car = car[car['Price'] != 'Ask For Price'] #Removing the row where value was "Ask for Price"
```

```python
# removing the commas and converting the dtype into integer
car['Price'] = car['Price'].str.replace(',','').astype(int)
```

```python
car.info()
```

```python
"""
Removing the commas and the word 'kms' by splitting them and getting the numeric value by 'str.get(0)'
and replacing commas with a space.
"""

car['kms_driven'] = car['kms_driven'].str.split(' ').str.get(0).str.replace(',','')
```

```python
# Keeping only numeric values and converting the dtype into integer
car = car[car['kms_driven'].str.isnumeric()]
```

```python
car['kms_driven'] = car['kms_driven'].astype(int)
```

```python
car.info()
```

```python
car = car[~car['fuel_type'].isna()] # '~' this symbol is for not as in this code it is neglecting NaN values
```

```python
"""
I want to keep the first 3 name of the car so firstly, i split the names by using 'str.split' 
then keeping the values by 'str.slice' and then join them by 'str.join'
"""

car['name'] = car['name'].str.split(' ').str.slice(0,3).str.join(' ')
```

```python
print("After CLEANING, the car dataset: ")
print()
car = car.reset_index(drop = True) #fixing the indexes as they are not in uniform state
car
```

```python
car.describe()
```

##### as shown above in desciptive stats, the 75% data has price under 5L but max value is around 85L so, let's check how many cars are more than 60L

```python
# As there is only one row where Price is more than 60L so we will remove.
car = car[car['Price']<6e6].reset_index(drop = True)
```

## MODEL

```python
# Assigning the variables for predictions and model training on the basis of target variable
X = car.drop(columns = 'Price')
y = car['Price'] #Target column
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
```

```python
ohe = OneHotEncoder()
ohe.fit(X[['name','company','fuel_type']])
```

```python
column_trans = make_column_transformer((OneHotEncoder(categories = ohe.categories_),
                                        ['name','company','fuel_type']),
                                       remainder = 'passthrough')
```

```python
lr = LinearRegression()
```

```python
pipe = make_pipeline(column_trans, lr)
```

```python
pipe.fit(X_train, y_train)
```

```python
y_pred = pipe.predict(X_test)
```

```python
print("R2_Score:")
r2_score(y_test,y_pred)
```

```python
scores = []
for i in range(1000):
    X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state = i)
    lr = LinearRegression()
    pipe = make_pipeline(column_trans, lr)
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    scores.append(r2_score(y_test,y_pred))
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state = np.argmax(scores))
lr = LinearRegression()
pipe = make_pipeline(column_trans, lr)
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)
r2_score(y_test,y_pred)
```

```python
import streamlit as st
import pickle
import numpy as np

# Load model
model = pickle.load(open('LinearRegressionModel.pkl', 'rb'))

st.title("🚗 Car Price Prediction App")

# Input fields
year = st.number_input("Manufacturing Year", 2000, 2025)
present_price = st.number_input("Present Price (in lakhs)")
kms_driven = st.number_input("Kilometers Driven")
fuel_type = st.selectbox("Fuel Type", ["Petrol", "Diesel"])
seller_type = st.selectbox("Seller Type", ["Dealer", "Individual"])
transmission = st.selectbox("Transmission", ["Manual", "Automatic"])
owner = st.number_input("Number of Owners", 0, 3)

# Convert categorical to numeric (example)
fuel_type = 0 if fuel_type == "Petrol" else 1
seller_type = 0 if seller_type == "Dealer" else 1
transmission = 0 if transmission == "Manual" else 1

# Prediction button
if st.button("Predict Price"):
    input_data = np.array([[year, present_price, kms_driven,
                            fuel_type, seller_type,
                            transmission, owner]])

    prediction = model.predict(input_data)
    st.success(f"Estimated Car Price: ₹ {prediction[0]:.2f} lakhs")
```
